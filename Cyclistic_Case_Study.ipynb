### Expert Step 7b: Memory Optimization & Schema Upgrades

To ensure our environment operates at peak performance and doesn't crash during complex calculations, we downcast data types. String columns with low cardinality (`rideable_type`, `member_casual`) are converted to the memory-efficient `category` type, coordinate floats are downcast to `float32`, and datetimes are parsed early.

```python
import os
import pandas as pd
import numpy as np

# Safely load combined_df from disk if not in memory
if 'combined_df' not in locals() and 'combined_df' not in globals():
    combined_file_path = '/content/drive/MyDrive/Cyclistic_Capstone/trips_combined.csv'
    print(f"Loading dataset from disk: {combined_file_path}...")
    combined_df = pd.read_csv(combined_file_path)

# Display memory footprint before optimization
mem_before = combined_df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"DataFrame Memory Before: {mem_before:.2f} MB")

# 1. Enforce Datetime Data Types
combined_df['started_at'] = pd.to_datetime(combined_df['started_at'])
combined_df['ended_at'] = pd.to_datetime(combined_df['ended_at'])

# 2. Downcast Coordinates to Float32
coord_cols = ['start_lat', 'start_lng', 'end_lat', 'end_lng']
for col in coord_cols:
    combined_df[col] = combined_df[col].astype(np.float32)

# 3. Categorize low-cardinality string columns
categorical_cols = ['rideable_type', 'member_casual']
for col in categorical_cols:
    combined_df[col] = combined_df[col].astype('category')

# Display memory footprint after optimization
mem_after = combined_df.memory_usage(deep=True).sum() / (1024 ** 2)
print(f"DataFrame Memory After: {mem_after:.2f} MB")
print(f"Saved Memory: {100 - (mem_after / mem_before * 100):.2f}%")

# Re-assign to df_full so all subsequent intermediate cells use this memory-optimized version
df_full = combined_df
```

### Step 31: Detect GPS Dispersion by Station

### Step 32: Detect Coordinates Ranges (Max - Min) by Station

To identify individual extreme coordinate outliers per station, we will calculate the span (difference between maximum and minimum value) for both latitude and longitude per starting station.

```python
# Step 32: Calculate coordinate ranges (Max - Min) per start station to catch single-trip outliers
coord_ranges = df_full[df_full['start_station_name'].notna()].groupby('start_station_name').agg(
    lat_min=('start_lat', 'min'),
    lat_max=('start_lat', 'max'),
    lng_min=('start_lng', 'min'),
    lng_max=('start_lng', 'max'),
    trip_count=('ride_id', 'count')
).reset_index()

coord_ranges['lat_range'] = coord_ranges['lat_max'] - coord_ranges['lat_min']
coord_ranges['lng_range'] = coord_ranges['lng_max'] - coord_ranges['lng_min']

# Sort by largest coordinate span to find potential anomalies
outlier_stations = coord_ranges[coord_ranges['trip_count'] >= 5].sort_values(by=['lat_range', 'lng_range'], ascending=False)
display(outlier_stations.head(10))
```

### Step 33: Detect End Coordinate Ranges (Max - Min) by Destination Station

We will calculate the span (Max - Min) for both ending latitude and ending longitude grouped by `end_station_name` to flag geographic anomalies on destination records.

```python
# Step 33: Calculate coordinate ranges (Max - Min) per end station to catch single-trip outliers
end_coord_ranges = df_full[df_full['end_station_name'].notna()].groupby('end_station_name').agg(
    lat_min=('end_lat', 'min'),
    lat_max=('end_lat', 'max'),
    lng_min=('end_lng', 'min'),
    lng_max=('end_lng', 'max'),
    trip_count=('ride_id', 'count')
).reset_index()

end_coord_ranges['lat_range'] = end_coord_ranges['lat_max'] - end_coord_ranges['lat_min']
end_coord_ranges['lng_range'] = end_coord_ranges['lng_max'] - end_coord_ranges['lng_min']

# Sort by largest coordinate span to find potential destination anomalies
outlier_end_stations = end_coord_ranges[end_coord_ranges['trip_count'] >= 5].sort_values(by=['lat_range', 'lng_range'], ascending=False)
display(outlier_end_stations.head(10))
```

### Step 34: Isolate and Count Geographic Outliers

We will calculate the median coordinate for each starting station as its true baseline location, then flag and count all trips where the actual start coordinates deviate by more than 0.05 degrees (~5.5 km) from that station's median baseline.

```python
# Step 34: Calculate station medians and count trips with significant coordinate deviation (> 0.05 degrees)
station_centroids = df_full[df_full['start_station_name'].notna()].groupby('start_station_name').agg(
    median_lat=('start_lat', 'median'),
    median_lng=('start_lng', 'median')
).reset_index()

# Merge medians back to the main dataset
df_merged = df_full.merge(station_centroids, on='start_station_name', how='left')

# Identify trips where the starting location is more than 0.05 degrees (~5.5 km) away from the station centroid
geographic_outliers = df_merged[
    (df_merged['start_station_name'].notna()) &
    ((abs(df_merged['start_lat'] - df_merged['median_lat']) > 0.05) |
     (abs(df_merged['start_lng'] - df_merged['median_lng']) > 0.05))
]

print(f"Total trips with severe start coordinate anomalies (> 0.05 deg deviation): {len(geographic_outliers)}")
if len(geographic_outliers) > 0:
    display(geographic_outliers[['ride_id', 'start_station_name', 'start_lat', 'median_lat', 'start_lng', 'median_lng', 'rideable_type', 'started_at']].head(10))
```

### Step 35: Audit for Absolute Zero or Extraneous Coordinates

We will check if any coordinates are set to exactly `0.0` (frequent default placeholders for GPS errors) or fall completely outside a logical regional boundary.

```python
# Step 35: Detect absolute zero coordinates or regional coordinate outliers
zero_coords = df_full[
    (df_full['start_lat'] == 0.0) | (df_full['start_lng'] == 0.0) |
    (df_full['end_lat'] == 0.0) | (df_full['end_lng'] == 0.0)
]

# Broad Illinois regional bounds: Lat [36.9, 42.6], Lng [-91.6, -87.0]
regional_outliers = df_full[
    ((df_full['start_lat'] < 36.9) | (df_full['start_lat'] > 42.6) |
     (df_full['start_lng'] < -91.6) | (df_full['start_lng'] > -87.0)) |
    ((df_full['end_lat'].notna()) &
     ((df_full['end_lat'] < 36.9) | (df_full['end_lat'] > 42.6) |
      (df_full['end_lng'] < -91.6) | (df_full['end_lng'] > -87.0)))
]

print(f"Trips with exact 0.0 coordinates: {len(zero_coords)}")
print(f"Trips outside the Illinois regional boundaries: {len(regional_outliers)}")
```

### Step 36: Identify Station Name Spelling & Casing Variations

We will check if there are station names that are identical when ignoring case, spaces, or trailing dots (e.g., 'Avenue' vs 'Ave'), which could cause grouping duplicates in down-stream analysis.

```python
# Step 36: Standardize names temporarily to find casing or spacing duplicate groups
unique_names = pd.Series(df_full['start_station_name'].dropna().unique())

# Create a DataFrame to map standardized names back to actual names
mapping_df = pd.DataFrame({
    'original_name': unique_names,
    'standardized_name': unique_names.str.strip().str.lower().str.replace('.', '', regex=False)
})

# Find standardized names that match to multiple original names
duplicate_groups = mapping_df.groupby('standardized_name').filter(lambda x: x['original_name'].nunique() > 1)

print(f"Total unique raw start station names: {len(unique_names):,}")
print(f"Number of spelling/casing variations detected: {len(duplicate_groups['standardized_name'].unique()):,}")
if len(duplicate_groups) > 0:
    display(duplicate_groups.sort_values(by='standardized_name'))
```

### Step 37: Identify Destination Station Name Spelling & Casing Variations

We will check if there are destination station names (`end_station_name`) that are identical when ignoring case, spacing, and abbreviations (like 'St.' vs 'St').

```python
# Step 37: Standardize destination names temporarily to find casing or spacing duplicates
unique_end_names = pd.Series(df_full['end_station_name'].dropna().unique())

# Create mapping DataFrame
mapping_end_df = pd.DataFrame({
    'original_name': unique_end_names,
    'standardized_name': unique_end_names.str.strip().str.lower().str.replace('.', '', regex=False)
})

# Find duplicates
duplicate_end_groups = mapping_end_df.groupby('standardized_name').filter(lambda x: x['original_name'].nunique() > 1)

print(f"Total unique raw end station names: {len(unique_end_names):,}")
print(f"Number of spelling/casing variations detected: {len(duplicate_end_groups['standardized_name'].unique()):,}")
if len(duplicate_end_groups) > 0:
    display(duplicate_end_groups.sort_values(by='standardized_name'))
```

### Step 38: Final Clean Record Count for Analysis

To wrap up our 38-step quality audit, we will exclude trips with negative durations and trips with missing end coordinates to get the final volume of 100% clean and usable records.

```python
# Step 38: Filter for clean dataset
# Exclude negative durations (< 0s) and missing end coordinates
clean_df = df_full[(df_full['duration_sec'] >= 0) & (df_full['end_lat'].notna()) & (df_full['end_lng'].notna())]

removed_count = len(df_full) - len(clean_df)
print(f"Initial raw records: {len(df_full):,}")
print(f"Records excluded (negative durations / missing coords): {removed_count:,}")
print(f"Final 100% clean and usable trips: {len(clean_df):,}")
print(f"Data retention rate: {len(clean_df) / len(df_full) * 100:.4f}%")
```
